```python
import pandas as pd
```

```python
data = pd.read_csv(f"../data/Telco_Customer_Churn.csv")
```

```python
#dropping duplicate rows and inplace=True means it apply changes in main data variable also
data.drop_duplicates(inplace=True)
data.duplicated().value_counts()
data.drop(columns=['customerID', 'gender'],inplace=True) #getting error bcz lready runned once so coulmns is not present in memory
#now total columns is 19 out of 21
```

```python

data.duplicated().value_counts() 
#value_counts() shows how many false and true count is their means total duplcated and non duplicated
```

```python
data.isna().sum()
data.dropna(inplace=True)#removes missing values
data.drop_duplicates(inplace=True)
data.duplicated().value_counts() 
```

```python
data["TotalCharges"] = pd.to_numeric(data["TotalCharges"], errors='coerce') #converting column to numeric from str
data.dropna(inplace=True)
```

```python
#Random forest work only on numerical data so we have to convert text to numers 

#converted [No internet service, No phone service, no]->0 yes->1
data.replace(["No internet service","No phone service","No","Yes"],[0,0,0,1],inplace=True)
data.to_csv("Cleaned_Customer_churn.csv")
```
